# 📊 Superstore Sales Data Analysis

## Project Overview
This project analyzes Superstore sales data to identify sales trends, customer behavior, product performance, and business insights using Python. The analysis focuses on sales, profit, regional performance, and customer segments to provide data-driven recommendations.

## Objectives

- Analyze total sales and profit.
- Identify top-performing products and customers.
- Analyze category-wise and region-wise performance.
- Understand monthly sales trends.
- Discover loss-making products and profit patterns.

## Tools Used
- Python
- Pandas
- Matplotlib
- Jupyter Notebook

## Dataset
The dataset contains information about orders, customers, products, sales, profit, discounts, and shipping details.




### 1.Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### 2.Load Dataset

```python
df = pd.read_csv(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\data\Sample - Superstore.csv",encoding="latin1")
print(df.head())
```

### 3.Data Understanding

```python
print("Rows and Columns:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nInformation:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Values:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
print(df.describe())
```

### 4.Data Cleaning

```python
# df = df.drop(columns=["Row ID","Country","Postal Code"]) --completed

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month_name()

df[["Order Date", "Year", "Month"]].head()
```

### 5.KPI Analysis

```python
# Total sales
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:",total_sales)
print("Total Profit:",total_profit)
```

### 6.Category-wise Sales

```python
category_sales = df.groupby("Category")["Sales"].sum()
category_sales.plot(kind="bar",figsize=(8, 5))

plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 7.Top 10 Products

```python
Top_product = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
Top_product.plot(kind="bar",
                 figsize=(10,5))
plt.title("Top 10 Product")
plt.xticks(rotation=45)


plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")
plt.show()
```

### 8.Monthly Sales Trend

```python
month_order = [
    "January", "February", "March",
    "April", "May", "June",
    "July", "August", "September",
    "October", "November", "December"
]

monthly_sales = (
    df.groupby("Month")["Sales"]
      .sum()
      .reindex(month_order)
)

monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(12, 5)
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 9.Region-wise Sales

```python
region_sales = df.groupby(
    "Region"
)["Sales"].sum()

region_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)

plt.title("Region Wise Sales")
plt.ylabel("")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 10.Sales vs Profit

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Sales"],
    df["Profit"]
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 11.Top 10 Customers

```python
Top_product = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
Top_product.plot(kind="bar",
                 figsize=(10,5))
plt.title("Top 10 Customers")
plt.xticks(rotation=45)
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 12.Category Wise Profit

```python
Category_profit = df.groupby("Category")["Profit"].sum()

Category_profit.plot(kind="bar")
plt.title("Category Wise Profit")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")
plt.show()
```

### 13.Region Wise Sales Contribution

```python
region_sales = df.groupby("Region")["Sales"].sum()
# print(region_sales)

region_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Region Wise Sales Contribution")
plt.ylabel("")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 14.Sub-Category Wise Sales

```python
sub_category_sales = df.groupby("Sub-Category")["Sales"].sum()

sub_category_sales.sort_values().plot(
    kind="barh",
    figsize=(8,6)
)

plt.title("Sub-Category Wise Sales")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

### 15.Sub-Category Wise Profit

```python
sub_category_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

sub_category_profit.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Sub-Category Wise Profit")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.savefig(r"C:\Users\utkarsh patil\OneDrive\Documents\Superstore Data Analysis\images.png")

plt.show()
```

# Business Insights

1. Technology category generated the highest sales.

2. Sales peak in November and December.

3. Higher discounts reduce profitability.

4. West region contributes the highest revenue.

5. Some products are generating losses and need review.

6. Top customers contribute significantly to total sales.

# Recommendations


1. Increase investment in the Technology category, as it generates the highest sales.

2. Review the pricing strategy for high-discount products to improve profitability.

3. Investigate loss-making products and identify the reasons behind their poor performance.

4. Increase the marketing budget during peak seasons (November and December) to maximize sales.

5. Introduce a loyalty program for top customers to improve customer retention and increase repeat purchases.

# Conclusion

This analysis highlights key sales trends, customer behavior, and profitability patterns. The insights and recommendations can help the company optimize its pricing strategy, improve customer retention, and maximize revenue growth.
